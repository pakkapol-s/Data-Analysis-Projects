## Load the 10-K filings

```python
ten_k_filings = get_filings(form='10-K')
ten_k_filing = ten_k_filings[0]
ten_k_report = ten_k_filing.obj()
```

## Access Financial Statements

```python
financial_data = ten_k_report.financials
financial_data
```

## Explore Financial Data

```python
# Access the income statement
income_statement = financial_data.income_statement

# Access the balance sheet
balance_sheet = financial_data.balance_sheet

# Access the statement of cash flows
cash_flow = financial_data.cashflow_statement()
```
