## Get all 10-K filings

```python
# 1. Import necessary functions and modules
from edgar import Company
from datetime import date
from dateutil.relativedelta import relativedelta # This is a new import for better date handling

# 2. Define the company
apple = Company("AAPL")

# 3. Get all filings for the company first, as the date filtering is a separate step
all_filings = apple.get_filings()

# 4. Filter the filings for 10-K and the desired date range (last 10 years)
# The `filter()` method is part of the Filings object
# Note the syntax: 'YYYY-MM-DD:YYYY-MM-DD' is used for date ranges
end_date = date.today()
start_date = end_date - relativedelta(years=10) # A more robust way to get 10 years ago

ten_k_filings = all_filings.filter(
    form="10-K", 
    filing_date=f"{start_date.isoformat()}:{end_date.isoformat()}"
)

# 5. Loop through the filtered filings to get the financial data
all_ten_k_data = []

for filing in ten_k_filings:
    try:
        ten_k_report = filing.obj()
        financial_data = ten_k_report.financials

        all_ten_k_data.append({
            'filing_date': filing.filing_date,
            'income_statement': financial_data.income_statement,
            'balance_sheet': financial_data.balance_sheet,
            'cash_flow_statement': financial_data.cashflow_statement()
        })
    except Exception as e:
        print(f"An error occurred while processing {filing.form} dated {filing.filing_date}: {e}")

print(f"Successfully processed {len(all_ten_k_data)} 10-K filings.")
```

## Save data as csv files

```python
import pandas as pd
# ... [Your existing code to get all_ten_k_data] ...

# 1. Combine all income statements into a single DataFrame
all_income_statements = pd.DataFrame()

for filing in all_ten_k_data:
    # First, call the method to get the Statement object
    statement_obj = filing['income_statement']()

    # Now, explicitly convert the Statement object to a pandas DataFrame
    df = statement_obj.to_dataframe()

    # Add the 'filing_date' column to the new DataFrame
    df['filing_date'] = filing['filing_date']

    # Concatenate the DataFrame
    all_income_statements = pd.concat([all_income_statements, df])

# 2. Save the combined DataFrame to an Excel file
all_income_statements.to_excel('all_apple_10k_income_statements.xlsx', index=False)

print("All income statements saved to all_apple_10k_income_statements.xlsx")

# Repeat for balance sheets and cash flows
# ...
```
