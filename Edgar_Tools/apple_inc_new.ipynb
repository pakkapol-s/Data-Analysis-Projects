## Get filings

## How to get filings:
1. Declare a variable to get filings as .get_filings()
2. Filter the desired filings as .filter(form = ' ', date = ' ')
3. Make that filing into a obj = .obj()
4. Access financial documents using .financials
5. There are 3 types of financial documents 
- company's balance sheet = financials.balance_sheet()
- company's income statement = financials.income_statement()
- company's cash flow statement = financials.cash_flow_statement()

```python
# Get company filings
filings = company.get_filings()
```

## Get all 10-K filings

Balance_Sheets

```python
# Filter the desired filings
annual_filings = filings.filter(form= '10-K', date = "2014-10-27:")

# Create an empty list to store each balance sheet DataFrame
balance_sheets_list = []

# Loop through each filing in the filtered collection
for filing in annual_filings:
    # 1. Get the data object for the current filing
    filing_report = filing.obj()

    # 2. Get the balance sheet (which is a Statement object)
    balance_sheet = filing_report.financials.balance_sheet()

    # 3. Convert the Statement object to a pandas DataFrame
    df = balance_sheet.to_dataframe()

    # 4. Add the filing date as a new column for reference
    df['filing_date'] = filing.filing_date

    # 5. Append the processed DataFrame to your list
    balance_sheets_list.append(df)

# Concatenate all DataFrames in the list into a single DataFrame
all_balance_sheets = pd.concat(balance_sheets_list, ignore_index=True)

# View the combined DataFrame
print(all_balance_sheets.head())

# Save the final result to an Excel file
all_balance_sheets.to_excel('apple_10k_balance_sheets_2014-2024.xlsx', index=False)

print("All balance sheets saved to apple_10k_balance_sheets_2014-2024.xlsx")
```
