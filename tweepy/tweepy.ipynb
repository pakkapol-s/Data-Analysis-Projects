```python
# read configs
config = configparser.ConfigParser()
config.read('config.ini')

bearer_token = config['twitter']['bearer_token']
```

```python
# Initialize the V2 Client with Bearer Token
client = tweepy.Client(bearer_token) 
```

```python
# STEP 2: Define your Search Query
# Define the keywords for the sentiment analysis project.
# The query is what will search public tweets, NOT your personal timeline.
search_query = 'Apple Vision Pro OR #VisionPro -is:retweet lang:en'
```

```python
# STEP 3: Use the Search Endpoint
try:
    # Use search_recent_tweets, which is a V2 endpoint and should be allowed.
    response = client.search_recent_tweets(
        search_query, 
        max_results=100, # Request up to 100 tweets
        tweet_fields=["created_at", "public_metrics"]
    )
    
    # The tweets are now in the 'data' attribute of the response object
    public_tweets = response.data
    
    if public_tweets:
        print(f"Successfully pulled {len(public_tweets)} public tweets.")
        print("First tweet text:", public_tweets[0].text)
    else:
        print("Search successful, but no tweets were found for the query.")

except tweepy.Forbidden as e:
    print(f"Error: Forbidden access even on V2 endpoint. Details: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
